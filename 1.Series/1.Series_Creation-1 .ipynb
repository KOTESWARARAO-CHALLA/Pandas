```python
import pandas as pd
```

```python
# using list
marks=[10,20,30,40,50]
ob=pd.Series(marks)
print(ob)
```

```python
# using tuple
marks=(10,20,30,40,50)
ob=pd.Series(marks,name="Marks",dtype="float")
print(ob)
```

```python
# using list with index
marks=[10,20,30,40,50]
id=["s1","s2","s3","s4","s5"]
ob=pd.Series(marks,index=id,name="Marks",dtype="int")
print(ob)
```

```python
# using dictionary
std={
    "S1":"10","S2":"20","S3":"30","S4":"40","S5":"50"
}

ob=pd.Series(marks,index=id,name="Marks",dtype="int")
print(ob)
# here keys becomes index and values becomes actuall values
```

```python
# Creating a Series Using Scalar Values
# A scalar is a single value, such as an integer, float, or string.
# We can create a Series by repeating a scalar value across multiple indexes.
# Suppose an e-commerce company wants to initialize the stock count for multiple warehouse locations.
stock=100
stores=["Balanagar","Chintal","kukatpally","Kompally","Bowenpally"]
ob=pd.Series(stock,index=stores,name="stock")
print(ob)
```

```python
# . Creating a Series Using NumPy Arrays
import numpy as np 
arr=np.array([1,2])
ob=pd.Series(arr)
print(ob)
```

```python
# using range
a=pd.Series(range(1,9))
print(a)
```
