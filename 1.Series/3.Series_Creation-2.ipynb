```python
# 1.create a series from "CSV FILE"
import pandas as pd 
std=pd.read_csv("students.csv")
ser=std.set_index("student_id")
series=pd.Series(ser["name"])
print(series)
```

```python
# 2.create a series from "Excel FILE"
emp=pd.read_excel("employees.xlsx")
empp=pd.Series(emp["employee_name"])
print(empp)
```

```python
# 3.create a series from "API"
import requests
url="https://dummyjson.com/users"
data=requests.get(url)
data=data.json()
users = pd.Series([user["firstName"] for user in data["users"]])
print(users)
```
