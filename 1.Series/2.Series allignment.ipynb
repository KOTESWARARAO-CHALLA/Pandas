```python
# 1. What is Series-to-Series alignment?
'''Series-to-Series alignment means Pandas matches values from two Series using 
their index labels rather than their positions.'''

# When you perform operations between two Series, 
# Pandas checks their index labels and matches the corresponding values.
# Series alignment is based on index labels, not the physical position of values.
```

```python
import pandas as pd
marks=pd.Series(
    [10,20,30,40,50,60],
    index=[1,2,3,4,5,6]
)
mark=pd.Series(marks,index=[6])
print(mark)
```

```python
import pandas as pd
marks = pd.Series(
    [85, 90, 75],
    index=["Ravi", "Kiran", "Arun"]
)

attendance = pd.Series(
    [95, 80, 90],
    index=["Ravi", "Kiran", "Sai"]
) 

result = marks + attendance

print(result)
```

```python
# Why did NaN appear?

# Pandas creates a combined index containing all unique labels from both Series.
# - Ravi: exists in both Series, so addition is possible.
# - Kiran: exists in both Series, so addition is possible.
# - Arun: missing from attendance, so result is NaN.
# - Sai: missing from marks, so result is NaN.
# Why does Pandas produce NaN instead of adding the available value?

# Because addition requires both operands. Pandas does not assume that a missing value is zero.
```

```python
# ---fix this
# We can handle this using methods such as reindex(), fillna(), 
# or arithmetic methods like add() with fill_value. 
# If we intentionally want positional operations instead of label-based alignment,
# we can convert Series into NumPy arrays using to_numpy().
```

```python
import pandas as pd

std = pd.read_csv("students.csv")

student_id = std["student_id"]
name = std["name"]
result = pd.Series(name, index=student_id, name="std_names")

print(result)
```

```python
# Here is what Pandas understands:
# - Data argument: name — index labels are 0 to 7.
# - New index: student_id — values are 101 to 108.
# Pandas tries to find index labels 101, 102, 103, etc. inside the name Series.

# But those labels don't exist.

# Therefore, you get NaN for every row.

# 4. How to fix it correctly
# There are three useful approaches.
```

```python
# Approach A: Use set_index() — useful in real projects 
# set_index() is an important Pandas DataFrame method used to make an existing column the index of a DataFrame.
std_indexed = std.set_index("student_id")

student_names = std_indexed["name"]

print(type(student_names)) 
# This is particularly useful when you want the student ID to become the actual DataFrame index.
```

```python
# Approach B: Use to_numpy() for positional assignment
result = pd.Series(
    name.to_numpy(),
    index=student_id,
    name="std_names"
)

print(result)
# This works because NumPy values do not carry Pandas index labels. 
# Pandas assigns the names according to their positions.
```
