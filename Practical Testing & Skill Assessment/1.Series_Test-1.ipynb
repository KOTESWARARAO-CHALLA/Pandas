```python
# Test scenario: E-commerce sales data analysis
# Task 1. Create daily sales data
import pandas as pd 
import numpy as np
daily_revenue=pd.Series(
    [12000,15000,18000,9000,22000],
    index=["Monday","Tuesday","Wednesday","Thursday","Friday"],
    name="daily_revenue"
)
print(daily_revenue)
```

```python
# Task 2. Create product inventory data 
products_data={
    "P101":50,
    "P102":120,
    "P103":30,
    "P104":75,
    "P105":200
}
products=pd.Series(products_data,name="products")
print(products) 
```

```python
# Task 3. Create customer ratings
ratings_array=np.array([4.5, 3.8, 5.0, 4.2, 3.5])
customer_IDs=["C001","C002","C003","C004","C005"]
ratings=pd.Series(ratings_array,customer_IDs,name="customer_ratings")
print(ratings)
```

```python
# Task 4. Create employee salary data 
employee_salary=pd.Series(
    [35000,45000,55000,40000,60000],
    index=["Rahul","Priya","Arjun","Sneha","Kiran"],
    name="employee_salary"
)
print(employee_salary["Priya"])
print(employee_salary["Kiran"])

```

```python
# Task 5. Retrieve specific sales records
# Using the daily sales Series from Task 1:
# 1. Retrieve Monday's revenue.
# 2. Retrieve Friday's revenue.
# 3. Retrieve Wednesday's revenue.
# 4. Retrieve Tuesday and Thursday's revenue individually.
print(daily_revenue["Monday"])
print(daily_revenue["Friday"])
print(daily_revenue["Wednesday"])
print(daily_revenue["Tuesday"])
print(daily_revenue["Thursday"])
```

```python
# Task 6. Customer data retrieval (10 marks)
# Using the customer ratings Series from Task 3:
# 1. Retrieve the rating of C003 using its label.
# 2. Retrieve the first customer's rating using its position.
# 3. Retrieve the last customer's rating using its position.
# 4. Retrieve the third customer's rating using positional indexing.
print(ratings["C003"])
print(ratings.iloc[0])
print(ratings.iloc[-1])
print(ratings.iloc[2])
```

```python
# Task 7. Warehouse stock investigation
# Using the inventory Series from Task 2:

# The warehouse manager asks you to investigate three products.
# 1. Retrieve the stock of P102 using its label.
# 2. Retrieve the stock of P104 using its position.
# 3. Retrieve the first three product records using positional slicing.
# 4. Retrieve the last two product records using positional slicing.
# 5. Retrieve P101, P103 and P105 using their labels.
print(products.loc["P102"])
print(products.iloc[3])
print(products.iloc[:3])
print(products.iloc[-2:])
print(products.loc[["P101","P103","P105"]])
```

```python
# Task 8. Employee salary investigation
The HR department wants to investigate employee salaries.

# Using Task 4:
# 1. Retrieve the salary of Arjun using label-based indexing.
# 2. Retrieve the second employee's salary using position-based indexing.
# 3. Retrieve the first three employees' salary records.
# 4. Retrieve the last two employees' salary records.
print(employee_salary["Arjun"])
print(employee_salary.iloc[1])
print(employee_salary.iloc[:3])
print(employee_salary.iloc[-2:])
```
