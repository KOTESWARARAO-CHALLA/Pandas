```python
import pandas as pd 

sales=pd.Series(
    [45000, 52000, 48000, 61000, 75000,
     68000, 82000, 90000, 85000, 95000],

    index=[
        "January", "February", "March", "April", "May",
        "June", "July", "August", "September", "October"
    ],
    name="Monthly Sales"
)
print(sales)
```

```python
# Label or CustomBased Indexing
print(sales[0]) #Error beacuse no label of 0
print(sales["January"]) #45000
```

```python
#Label Based Indexing - .loc[]
print(sales.loc["January"])
print(sales.loc[["January", "February"]]) #Access more than one elements
```

```python
# Position-Based Indexing — .iloc[] 
print(sales.iloc[0])
print(sales.iloc[[0,1,2,4]]) #Access more than one elements
# print(sales[0]) #fix this one by above
```

```python
#Slicing
```

```python
print(sales["January":"April"]) #label-based slicing includes the ending label when it exists.
print(sales.loc["January":"April"]) #label-based slicing includes the ending label when it exists
print(sales.iloc[:5]) #Position-based slicing includes position 0 but excludes position 5
```
