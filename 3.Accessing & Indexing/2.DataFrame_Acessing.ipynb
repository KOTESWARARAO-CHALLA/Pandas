```python
selecting specific rows and columns together.
2. Understanding loc[]
-----
loc[] is a label-based indexer.

It selects rows and columns using their labels.
Syntax
------
df.loc[row_label, column_label]

The first argument represents the row selection.

The second argument represents the column selection.
```

```python
std=pd.read_csv("../DataSets/student_marks_50_records.csv")
print(std.head())
```

```python
3. Understanding iloc[]
iloc[] is an integer-position-based indexer.

Unlike loc[], it does not use row or column names.

It uses integer positions.
Syntax
df.iloc[row_position, column_position]
```

```python
record1=std.iloc[1,7]
record2=std.iloc[:,0]
record3=std.iloc[:,[0,7]]
record4=std.iloc[1,:]
record5=std.iloc[[1,5,10],:]
record6=std.iloc[:5,0:8]
record7=std.iloc[1:10:2,:]
print(record7)
```

```python
Operation	       loc[]	                     iloc[]
Single row   	df.loc["P103"]          	df.iloc[2]
Single column	df.loc[:, "price"]	    df.iloc[:, 1]
Single value	df.loc["P103","price"]	df.iloc[2,1]
Multiple rows	df.loc[["P101","P103"]]	df.iloc[[0,2]]
Multiple columnsdf.loc[:,["price","stock"]]	df.iloc[:,[1,2]]
Rows and columns df.loc[["P101","P103"],["price"]]	df.iloc[[0,2],[1]]

Professional recommendation: 
Use loc[] when you know the names of rows and columns.
Use iloc[] when you want to select data based on positions.
```
