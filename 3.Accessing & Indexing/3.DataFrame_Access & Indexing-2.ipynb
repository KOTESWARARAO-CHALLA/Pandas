```python
import pandas as pd 
df = pd.read_csv("../DataSets/student_marks_50_records.csv")
print(df)
```

```python
1. Custom Index Using set_index()
1.1 What is a custom index?
By default, Pandas automatically assigns integer indexes to DataFrame rows.

Important methods to remember:
Method          	Purpose
set_index()-------Convert a column into an index
                  By default, set_index() returns a new DataFrame and does not change the original
                  customer_df = df.set_index("customer_id")
reset_index()-----Restore an index as a column
                  df.set_index("customer_id", inplace=True)


```

```python
std=df.set_index("stdname")
std=std.reset_index()
print(std)
```

```python
3. Accessing Individual Cells Using at[] and iat[]
Sometimes you do not need an entire row or column. You only need one particular value.

For this purpose, Pandas provides:
- at[] — Label-based scalar access.
         df.at[row_label, column_label]
- iat[] — Position-based scalar access.
         df.iat[row_position, column_position]
    
3.4 Comparison
Method	Access type	Example
loc[]	Label-based, one or more values	df.loc["P102", "price"]
iloc[]	Position-based, one or more values	df.iloc[1, 1]
at[]	Label-based, single value	df.at["P102", "price"]
iat[]	Position-based, single value	df.iat[1, 1]
```

```python
std=df.set_index("stdname")
print(std.columns)
r1=std.at["Kavya","total marks"]
r2=std.iat[0,6]
print(r2)
```

```python
Conditional Selection
Conditional selection means selecting DataFrame rows based on a particular condition.

For example:
- Select customers whose age is greater than 30.
- Select employees whose salary is above ₹50,000.
- Select houses whose price is below ₹80 lakhs.
- Select students whose marks are above 80.
Instead of manually accessing rows, we tell Pandas what condition the data must satisfy.
1. Basic comparison operators
    
Operator  Syntax example
==	  	 df[df["age"] == 25]
!=	     Not equal to	df[df["age"] != 25]
>      	 Greater than	df[df["salary"] > 50000]
<	     Less than	df[df["age"] < 30]
>=	     Greater than or equal to	df[df["salary"] >= 50000]
<=	     Less than or equal to	df[df["age"] <= 30]

2. Multiple conditions
Operator		                                       example
&	AND — both conditions must be True	        df[(df["age"] > 25) & (df["salary"] > 50000)]
\|	OR — at least one condition must be True	df[(df["age"] > 25) \| (df["salary"] > 50000)]
~	NOT — reverses the condition	            df[~(df["age"] > 25)]

Important: Always put each condition inside parentheses when using &, |, or ~.

4. Conditional selection using loc[]
Requirement	                                                Syntax
Select rows based on one condition	              df.loc[df["age"] > 25]
Select specific columns based on a condition	  df.loc[df["age"] > 25, ["name", "salary"]]
Select rows satisfying two conditions	          df.loc[(df["age"] > 25) & (df["salary"] > 50000)]
Select rows satisfying either condition	           df.loc[(df["department"] == "IT") \| (df["department"] == "HR")]
```
