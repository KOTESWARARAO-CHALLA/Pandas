```python
In Pandas, attributes provide information about a Series or DataFrame without modifying the actual data.

They help us understand:
- How much data is present.
- What type of data is stored.
- How many rows and columns exist.
- What the column names and indexes are.
- How much memory the data occupies.
- Whether the data is empty or not.

   Attribute	     Definition	                                      Use
df.index	     Returns row labels	                              Identify rows
df.columns	     Returns column labels	                          Identify columns
df.values	     Returns underlying values	                      Access data as an array
df.dtypes	     Returns data type of each column	              Understand column data types
df.shape	     Returns (rows, columns)	                      Understand dataset dimensions
df.size	         Returns total number of elements	              Count all data elements
df.ndim	         Returns number of dimensions	                  Check dimensionality
df.empty	     Returns True if either axis has length zero	  Check whether DataFrame is empty
df.T	         Returns transposed DataFrame	                  Exchange rows and columns
df.axes	         Returns row and column indexes	                  Inspect both axes
df.memory_usage()	Returns memory usage of each column	Analyze   memory consumption
df.select_dtypes()	Selects columns based on data types	Filter    numerical or categorical columns

For AI/ML, 
become comfortable with shape, size, ndim, index, columns, dtype, dtypes, values, empty, and memory_usage().
These are particularly useful when exploring datasets before machine learning.
```
