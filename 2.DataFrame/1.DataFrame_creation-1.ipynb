```python
import pandas as pd
# 1. Creating a DataFrame Using a Dictionary
data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Keyboard"],
    "Price": [60000, 25000, 30000, 2000],
    "Quantity": [5, 10, 7, 20]
}
df=pd.DataFrame(data)
print(df)
```

```python
# 2.Creating a DataFrame Using a List of Dictionaries
data = [
    {"Name": "Rahul", "Age": 24, "City": "Hyderabad"},
    {"Name": "Priya", "Age": 22, "City": "Bangalore"},
    {"Name": "Arun", "Age": 25, "City": "Chennai"}
]
df=pd.DataFrame(data)
print(df)
```

```python
# 3.Creating a DataFrame Using a List of Lists
data = [
    ["Rahul", 24, "Hyderabad"],
    ["Priya", 22, "Bangalore"],
    ["Arun", 25, "Chennai"]
]
df=pd.DataFrame(data,columns=["Name","Age","City"])
print(df)
```

```python
# Creating a DataFrame From Existing Pandas Series
products = pd.Series(["Laptop", "Mobile", "Tablet"])
prices = pd.Series([60000, 25000, 30000])
quantity = pd.Series([5, 10, 7])

df = pd.DataFrame({
    "Product": products,
    "Price": prices,
    "Quantity": quantity
})

print(df)
```

```python
# Creating a DataFrame Using NumPy Arrays
import numpy as np


data = np.array([
    [25, 50000],
    [30, 60000],
    [35, 75000]
])

df = pd.DataFrame(
    data,
    columns=["Age", "Salary"]
)

print(df)
```

```python
# Creating a DataFrame With a Custom Index 
data = {
    "Product": ["Laptop", "Mobile", "Tablet"],
    "Price": [60000, 25000, 30000]
}

df = pd.DataFrame(
    data,
    index=["P101", "P102", "P103"]
)

print(df)
```
